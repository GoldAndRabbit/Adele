```python
from nemo.collections.asr.models import EncDecMultiTaskModel

canary_model = EncDecMultiTaskModel.from_pretrained("nvidia/canary-1b-flash")

decode_cfg = canary_model.cfg.decoding
decode_cfg.beam.beam_size = 1
canary_model.change_decoding_strategy(decode_cfg)
```

```python
# wget https://isv-data.oss-cn.hangzhou.aliyuncs.com/ics/MaaS/ASR/test_audio/asr_sample_fr.wav
```

```python
output = canary_model.transcribe(
    audio=["./asr_example_fr.wav"],
    souce_lang="fr",
    target_lang="fr",
    pnc="yes",
    timestamps=True,
)
predicted_text = output[0]["text"]
word_level_timestamps = output[0]["word"]
segment_level_timestamps = output[0]["segment"]
```
